### Вопросы для анализа 

1. **Анализ индекса массы тела.** Людей с каким ИМТ больше всего? Есть ли люди, имеющие слишком низкий/высокий ИМТ, опасный для здоровья? Почему их ИМТ находится вне нормальных граници как это отражается на их психическом состоянии и продуктивности?  
2. **Анализ показателей сна.** Люди с каким хронотипом преобладают в данных? Правда ли «совы» спят меньше, а «жаворонки» продуктивнее? Есть ли те, кто спит аномально большое количество часов и почему? Есть ли те, кто спит мало и/или чей сон не является качественным и как это влияет на их работу, учебу и психическое состояние? Склонны ли люди, которые спят плохо, переедать и набирать вес? 
3. **Стабильность веса.** Есть ли те, чей вес сильно меняется на протяжении нескольких дней? Как это связано с питанием и стрессом у таких людей? 
4. **Набор и сброс веса.** Есть ли те, чей вес стабильно растет/падает? Они набирают вес/худеют умышленно или это происходит в связи с внешними обстоятельствами? 
5. **Тренировки и пульс.** Как тренировки влияют на пульс в спокойном состоянии? Есть ли те, чей пульс экстремально низкий, но при этом они не занимаются спортом активно? 
Отличный вызов. Связать задачи с весом и калориями — это классика аналитики здоровья. А раз ты просишь **максимально сложные задачи без ML**, нам придется копнуть в сторону продвинутой статистики, временных рядов (longitudinal analysis) и поиска скрытых аномалий.
6. **Влияние факторов с задержкой. Анализ временных рядов.** С какой задержкой факторы образа жизни влияют на вес? Через сколько дней после тяжелой тренировки и высокой крепатуры вес падает сильнее всего? Как быстро вес реагирует на стресс? Есть ли у людей «недельные циклы» — например, вес стабильно падает к пятнице и растет к понедельнику?
7. **Поиск аномалий.** Есть ли люди с ИМТ > 35, которые каждый день проходят более 15 000 шагов и тратят более 1000 ккал на тренировках? Как это объяснить? Есть ли люди с ИМТ < 20, которые стабильно потребляют более 3500 ккал в день? Существуют ли «суперстабильные» люди, чей вес абсолютно не меняется на протяжении 14 дней, несмотря на колебания калорий и активности?
8. **Анализ потребления калорий (Regression Decision Tree).** Какие факторы (стресс, сон, шаги, тренировки, ИМТ) сильнее всего влияют на суточное потребление калорий (calories_kcal)? Правда ли, что при высоком стрессе люди едят больше (заедают стресс), или наоборот — стресс убивает аппетит? Есть ли порог по шагам, после которого калории перестают расти?
9. **Анализ набора веса (Classification Decision Tree).** Какие паттерны поведения (например, «плохой сон + много кофеина + отсутствие тренировки») приводят к скачку веса на следующий день? Влияет ли вчерашний кофеин, гидратация или стресс на сегодняшний вес? Есть ли люди, чей набор веса предсказуем со 100% вероятностью, и что у них общего?
